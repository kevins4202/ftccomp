```python
import pandas as pd
import numpy as np
from bokeh.plotting import figure
from bokeh.models import ColumnDataSource
import csv
from bokeh.io import output_notebook
import os
# Regression
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
output_notebook()
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
broken = []
```

```python
with open(os.getcwd() + '/scripts/scores.csv', 'w') as out_file:
    writer = csv.writer(out_file)
    writer.writerow(('OPR1', 'Auto1', 'CCWM1', 'OPR2', 'Auto2', 'CCWM2', 'OPR3', 'Auto3', 'CCWM3', 'OPR4', 'Auto4', 'CCWM4','score1', 'score2'))
for d in os.listdir(os.getcwd() + '/scripts'):
    print(d)
    if 'matches-' in d:
        match = d.split('matches-')[1].split('.txt')[0]
        
#         print(match)
        
        with open(os.getcwd() + '/scripts/'+d, 'r') as in_file:
            stripped = (line.strip() for line in in_file)
            lines = (line.split(" ") for line in stripped if line)
            
            for l in list(lines):
                skip = False
                teams = l[0:2] + l[4:6]
                with open(os.getcwd() + '/scripts/scores.csv', 'a') as out_file:
                    writer = csv.writer(out_file)
                    
                    to_write = []
                    
                    for tm in teams:
                        try:
                            tm_df = pd.read_csv(os.getcwd()+ f'/data/team_data/{tm}.csv')
                        except FileNotFoundError:
                            skip = True
                            broken.append([match, tm])
                            break
                            
                        
                        
                        row = tm_df[tm_df['ID']==match]
                        
                        if(row.shape[0] ==0):
                            skip = True
                            broken.append([match, tm])
                            break
                        
#                         print(row)
                        
                        app = [row.iat[0,k] for k in range(1,4)]
                        
                        to_write += app
                    
                    to_write += [l[2], l[6]]
                    
                    if not skip:
                        writer.writerow(to_write)
                
                
```

```python
scores = pd.read_csv(os.getcwd() +'/scripts/scores.csv', delimiter = ",")
```

```python
scores
```

```python
x = scores[scores.columns[:-2]]
```

```python
y = scores[scores.columns[-2:]]
```

```python
y
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.3)
```

```python
model = LinearRegression()
```

```python
model.fit(x_train, y_train)
```

```python
print(model.intercept_)
```

```python
predictions = model.predict(x_test)
```
