```python
import datetime
all_df = renamed_df
for i in range(renamed_df.shape[0]):
    s = renamed_df.at[i,"Start"].split("-")
    e = renamed_df.at[i,"End"].split("-")
    all_df.at[i,"Start"] = datetime.datetime(int(s[0]),int(s[1]),int(s[2]))
    all_df.at[i,"End"] = datetime.datetime(int(e[0]),int(e[1]),int(e[2]))
```
