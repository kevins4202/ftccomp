```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import os
```

```python
matches = pd.read_csv(os.getcwd().replace('scripts', 'data')+ '/all_matches.csv')
```

```python
matches
```

```python
import time
from bs4 import BeautifulSoup
from selenium import webdriver

url = "https://ftc-events.firstinspires.org/2022/"
driver = webdriver.Safari()
all_teams = {}
for i in range(matches.shape[0]):
    try:
        add = matches.at[i, 'Link']
        driver.get(url + add)
        src = driver.page_source
        soup = BeautifulSoup(src, 'lxml')
        teams = soup.find('table', {'id': 'teamtable'}).find('tbody').find_all('a')
        names = soup.find('table', {'id': 'teamtable'}).find('tbody').find_all('td', {'class' : 'teamnamecol'})

        if(len(teams) != len(names)):
            print(i)
            break

        for i in range(len(teams)):
            all_teams[int(teams[i]['href'].split("/")[-1])] = names[i].text
    except Exception as e:
        continue
        
driver.close()
```

```python
all_teams
```

```python
all_teams[22802] = 'Swanage'
```

```python
df = pd.DataFrame(all_teams.items(), columns=['Team Number', 'Team Name'])
```

```python
df = df.rename({0: "Team Number"}, axis = "columns")
```

```python
df.to_csv(os.getcwd().replace('scripts', 'data') + '/teams.csv', index = False)
```
