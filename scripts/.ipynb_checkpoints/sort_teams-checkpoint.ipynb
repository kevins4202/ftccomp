```python
wlt = pd.concat([data["Team Number"], data["Event W-L-T"]], axis = 1)
```

```python
x = lambda a,b: (int(a.split("-")[0])+ int(b.split("-")[0])) +"-" +(int(a.split("-")[1])+ int(b.split("-")[1]))+"-" +(int(a.split("-")[2])+ int(b.split("-")[2]))
```

```python
w = []
l = []
t = []
for index, row in wlt.iterrows():
    arr = row["Event W-L-T"].split("‑")
    
    w.append(int(arr[0]))
    l.append(int(arr[1]))
    t.append(int(arr[2]))
    
wlt["W"] = w
wlt["L"] = l
wlt["T"] = t
```

```python
teams = noRP["Team Number"].unique()
```

```python
regions = []
names = []
for team in teams:
    regions.append(noRP[noRP["Team Number"]==team].iloc[[0]]["Region"].to_string(header=False, index=False))
    names.append(noRP[noRP["Team Number"]==team].iloc[[0]]["Team Name"].to_string(header=False, index=False))
```

```python
names
```

```python
team_info = pd.DataFrame({'Team Number' : teams,
                                'Team Name' : names,
                                'Region' : regions }, 
                                )
```

```python
team_info.to_csv(path +'team_info.csv', index = False)
```
