```python
w = []
l = []
t = []
for index, row in wlt.iterrows():
    arr = row["Event W-L-T"].split("‑")
    
    w.append(int(arr[0]))
    l.append(int(arr[1]))
    t.append(int(arr[2]))
    
wlt["W"] = w
wlt["L"] = l
wlt["T"] = t
```
